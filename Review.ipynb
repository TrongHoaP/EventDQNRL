```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
path = "runs\\approach_metrics_5step_demand03.csv"
```

```python
df = pd.read_csv(path)
```

```python
df.head()
```

```python
approach_id = "C1B1"
approach_df = df[df["approach_id"] == approach_id]
#get time from 40000 to 50000
approach_df = approach_df[(approach_df["time"] >= 60000) & (approach_df["time"] <= 65000)]
# line plot speed, occupancy, and jam_veh over time for the approach
plt.figure(figsize=(12, 6))
plt.plot(approach_df["time"], approach_df["speed"], label="Speed (m/s)")
# plt.plot(approach_df["time"], approach_df["occupancy"], label="Occupancy (%)")
# plt.plot(approach_df["time"], approach_df["jam_veh"], label="Jam Vehicles")
plt.xlabel("Time (s)")  
plt.ylabel("Value")
plt.title(f"Approach {approach_id} Metrics Over Time")
plt.legend()
plt.grid()
plt.show() 
```
